```python
api_key = input("Enter the api_key")
```

```python
import requests
import pyttsx3
import speech_recognition as sr
```

# speach to text

```python
recognizer = sr.Recognizer()

with sr.Microphone() as source:
    print("🎤 Speak the village name...")
    recognizer.adjust_for_ambient_noise(source, duration=1)
    audio = recognizer.listen(source, timeout=10, phrase_time_limit=5)

try:
    village = recognizer.recognize_google(audio, language="en-IN")
    village = village.strip()

    print("Recognized Village:", village)

except sr.UnknownValueError:
    print("Sorry, I could not recognize the village name.")

except sr.RequestError:
    print("Speech recognition service is unavailable.")

except sr.WaitTimeoutError:
    print("No speech detected. Please try again.")
```

# retriving the data from the server 

```python
url = f"https://api.openweathermap.org/data/2.5/weather?q={village},IN&appid={api_key}&units=metric"

response = requests.get(url, timeout=10)

if response.status_code == 200:
    data = response.json()

    temperature = data["main"]["temp"]
    wind_speed = data["wind"]["speed"] * 3.6
    weather = data["weather"][0]["description"]
    location = data["name"]

    print("\nWeather Report")
    print("-------------------------")
    print(f"Village/Location: {location}")
    print(f"Temperature: {temperature} °C")
    print(f"Wind Speed: {wind_speed:.2f} km/h")
    print(f"Weather Condition: {weather}")

else:
    print("Unable to find weather information for this village.")
    print("Please check the village name or API key.")
```

# text to speach 

```python
engine = pyttsx3.init()

message = (
    f"The temperature in {location} is {temperature} degrees Celsius. "
    f"The wind speed is {wind_speed:.2f} kilometres per hour. "
    f"The weather condition is {weather}."
)

print("\nAssistant:", message)

engine.say(message)
engine.runAndWait()
```
